```python
source_file = f'{landing_folder_path}/{v_batch_id}/constructors.json'
table_name = f'{catalog_name}.{bronze_schema}.constructors'
```

```python
# DDL approach for schema enforcement.
constuctors_schema = 'constructorId STRING, name STRING, nationality STRING, url STRING'
```

```python
constructors_df = (
    spark.read
    .format('json')
    .option('mode','FAILFAST')
    #.option('inferSchema','true')
    .schema(constuctors_schema) 
    .load(source_file)
)
```

```python
display(constructors_df)
```

```python
# Ingesting metadata
constructors_df_final = add_file_metadata(constructors_df)
```
