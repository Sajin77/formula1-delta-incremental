```python
# function to add the file metadata for ingestion

from pyspark.sql import functions as F

def add_file_metadata(df):
    return (
        df.withcolumns(
            {
                "ingestion_timestamp": F.current_timestamp(),
                "source_file": F.col('_metadata.file_path')
            }
        )
    )
        
    

    
```

```python
def write_to_bronze(
    input_df,
    target_table,
    new_batch_id
):
    final_df = input_df.withColumn("batch_id", F.lit(new_batch_id))
    (
        final_df
            .write
            .format('delta')
            .mode('overwrite')
            .partitionBy('batch_id')
            .option('overwriteSchema', 'true')
            .option('replaceWhere', f"batch_id = '{new_batch_id}'")
            .saveAsTable(target_table)
    )

```
