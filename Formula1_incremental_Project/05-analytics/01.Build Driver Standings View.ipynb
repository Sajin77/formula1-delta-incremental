# Build Driver Standings
#### Sources
 1. fact_session_results
 1. dim_drivers

#### Output Columns
 1. season
 1. driver id
 1. driver name
 1. nationality
 1. race starts
 1. total points
 1. number of wins
 1. number of podiums
 1. standing position


```sql
CREATE OR REPLACE VIEW formulae1.gold.v_driver_standing  -- creating driver tsanding view
AS (
    with driver_season_summary AS(  -- CTE
    SELECT r.season, 
        r.driver_id, 
        d.driver_name, 
        d.nationality, 
        COUNT(*) AS race_starts, 
        SUM(r.points) AS total_points,
        COUNT_if(r.is_win) AS number_of_wins,
        COUNT_if(r.is_podium) AS number_of_podiums
        FROM formulae1.gold.fact_session_results r
        JOIN formulae1.gold.dim_drivers d
        ON r.driver_id = d.driver_id
        GROUP BY r.season, 
        r.driver_id, 
        d.driver_name, 
        d.nationality 
)

SELECT   
    season,
    driver_id,
    driver_name,
    nationality,
    race_starts,
    RANK() OVER (PARTITION BY season ORDER BY total_points DESC, number_of_wins DESC) as standing_position,
    total_points,
    number_of_wins,
    number_of_podiums
FROM driver_season_summary
)
```

```sql
SELECT * FROM formulae1.gold.v_driver_standing;
```
